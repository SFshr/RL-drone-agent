```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
MASS = 0.066
YLENGTH = 0.085
OUTFILE = 'drive/MyDrive/RAINinertia'
#have to convert stl to voxels https://drububu.com/miscellaneous/voxelizer/?out=txt
vcoords = np.loadtxt('drive/MyDrive/RAINV5voxel.txt', unpack = True, delimiter=',')[:3]

#swap y and z axes
vcoords[[1,2]] = vcoords[[2,1]]

#get width of a voxel
maxy = max(vcoords[1])
miny = min(vcoords[1])
vlength = maxy - miny + 1
vwidth = YLENGTH/vlength

#center coordinates on center of mass
numpoints = len(vcoords[0])
xcoord = vcoords[0] - sum(vcoords[0])/numpoints
ycoord = vcoords[1] - sum(vcoords[1])/numpoints
zcoord = vcoords[2] - sum(vcoords[2])/numpoints

#compute moments of inertia and inertia tensor
xsquare = np.square(xcoord)
ysquare = np.square(ycoord)
zsquare = np.square(zcoord)

Ixx = sum(ysquare + zsquare)
Iyy = sum(xsquare + zsquare)
Izz = sum(xsquare + ysquare)
Ixy = -sum(xcoord * ycoord)
Ixz = -sum(xcoord * zcoord)
Iyz = -sum(ycoord * zcoord)

scalefactor = (MASS/numpoints) * (vwidth**2)
inertiatensor = np.matrix([[Ixx,-Ixy,-Ixz],[-Ixy,Iyy,-Iyz],[-Ixz,-Iyz,Izz]]) * scalefactor
np.save(OUTFILE,inertiatensor)
```
